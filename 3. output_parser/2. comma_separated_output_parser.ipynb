# Comma-Separated List Output Parser

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
chat = ChatOpenAI(
    model_name='llama3.2:3b',
    openai_api_key='ollama', 
    openai_api_base='http://localhost:11434/v1',
    temperature = 0, 
    max_tokens = 100,
    model_kwargs = {
        'seed':365
    }
)
```

```python
message_h = HumanMessage(
    content = f''' I've recently adopted a dog. Could you suggest some dog names? {CommaSeparatedListOutputParser().get_format_instructions()}'''
    )
```

```python
print(message_h.content)
```

```python
response = chat.invoke([message_h])
print(response.content)
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
response_parsed = list_output_parser.invoke(response)
```
